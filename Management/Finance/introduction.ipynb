# 1. Motivation

Finance is concerned with how individuals, businesses, and institutions make decisions about money and financial resources.

The study of finance is motivated by the need to make decisions involving:

- The allocation of financial resources
- Investment
- Financing
- Risk
- Time
- Uncertainty
- Value creation

# 2. Dramatis Personae

**Dramatis personae** means the main actors or participants involved in financial decisions.

Typical participants in finance include:

- **Individuals / Households:** Save, borrow, invest, and consume.
- **Businesses / Corporations:** Raise capital, invest in projects, and manage financial resources.
- **Financial Institutions:** Banks, insurance companies, investment funds, and other intermediaries.
- **Governments:** Collect taxes, spend public money, borrow, and manage public finances.
- **Investors:** Provide capital in exchange for an expected financial return.
- **Financial Markets:** Provide mechanisms through which financial assets and capital are exchanged.

# 3. Fundamental Challenges of Finance

Finance deals with several fundamental challenges:

### Resource Allocation
Financial resources are limited, so they must be allocated to their most valuable uses.

### Time
Financial decisions often involve receiving or paying money at different points in time.

### Risk and Uncertainty
Future outcomes are uncertain. Financial decisions therefore require the analysis and management of risk.

### Information
Financial decisions depend on the quality, availability, and interpretation of information.

### Conflicting Interests
Different participants may have different objectives, creating potential conflicts of interest.

# 4. Framework for Financial Analysis

A financial analysis framework provides a structured way to evaluate financial decisions.

A basic framework can be organized around:

1. **Cash flows** — What money comes in and what money goes out?
2. **Timing** — When do the cash flows occur?
3. **Risk** — How uncertain are the expected outcomes?
4. **Value** — What is the economic value of the decision?
5. **Alternatives** — What other choices are available?
6. **Decision** — Which alternative creates the most value relative to its risk?

This framework is used in areas such as investment analysis, corporate finance, and financial planning.

# 5. Importance of Time and Risk

Two fundamental dimensions of finance are **time** and **risk**.

## Time

The **time value of money** states that money available today is generally worth more than the same amount received in the future.

This is because money available today can potentially be invested and earn a return.

Important concepts include:

- Present Value (PV)
- Future Value (FV)
- Interest
- Compounding
- Discounting

## Risk

Risk refers to the uncertainty surrounding future financial outcomes.

An investment with uncertain future cash flows involves risk. Investors generally require compensation for taking additional risk.

A central idea in finance is the relationship between **risk and expected return**.

# 6. Six Principles of Finance

The six principles provide a foundation for understanding financial decision-making.

### Principle 1 — Money Has a Time Value
A unit of money today is generally more valuable than the same unit received in the future.

### Principle 2 — Risk Requires a Reward
People generally require compensation for accepting additional financial risk.

### Principle 3 — Information Is the Basis for Decisions
Financial decisions depend on relevant and reliable information.

### Principle 4 — Markets Determine Prices
Financial markets bring buyers and sellers together and help determine the prices of financial assets.

### Principle 5 — Incentives Matter
People respond to financial incentives, and incentives can influence financial decisions.

### Principle 6 — Financial Decisions Should Create Value
A good financial decision should increase economic value or provide an adequate return relative to the resources and risks involved.

# 8. How to Get the Most Out of This Course

To learn finance effectively:

1. **Understand the concepts before memorizing formulas.**
2. **Practice financial calculations regularly.**
3. **Understand what each variable in a formula represents.**
4. **Connect theory to real financial situations.**
5. **Practice interpreting results, not only calculating them.**
6. **Review mistakes and understand why they occurred.**
7. **Use examples to connect time, risk, return, and value.**

Finance becomes much easier when mathematical calculations are connected to the underlying economic reasoning.

# Summary

The introductory study of finance revolves around several fundamental ideas:

- Financial resources are limited.
- Financial decisions involve **time and risk**.
- Different participants have different financial objectives.
- Cash flows are central to financial analysis.
- The value of money depends on when it is received or paid.
- Risk and expected return are closely related.
- Financial decisions should ultimately be evaluated in terms of **value creation**.

These concepts form the foundation for more advanced topics in finance.
