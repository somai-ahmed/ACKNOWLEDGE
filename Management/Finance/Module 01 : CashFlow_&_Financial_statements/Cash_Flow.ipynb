# Free Cash Flow (FCF)
### Operating Cash Flow, Capital Spending & Change in Net Working Capital

> *A company can generate cash from its operations, but how much of that cash is actually available after investing in its assets and supporting its day-to-day operations?*

That question is the core of **Free Cash Flow (FCF)**.

A company generates cash through its operations, but it cannot keep all of it — it may need to:
- invest in equipment, buildings, and technology
- purchase inventory
- extend credit to customers
- maintain enough working capital to operate

We get from **Operating Cash Flow** to **Free Cash Flow** by accounting for these cash requirements.

---
## 1. Operating Cash Flow — OCF

**OCF** measures the cash generated by the company's normal operations. The standard corporate-finance formula is:

$$\boxed{OCF = EBIT(1-T) + \text{Depreciation}}$$

where **EBIT** = Earnings Before Interest and Taxes, **T** = tax rate, and **Depreciation** is added back because it reduces accounting profit but is *not* a current cash payment.

**Example**

| Item | Amount |
|---|---:|
| EBIT | \$30,000 |
| Tax rate | 25% |
| Depreciation | \$5,000 |

````python
EBIT = 30_000
tax_rate = 0.25
depreciation = 5_000

after_tax_ebit = EBIT * (1 - tax_rate)
OCF = after_tax_ebit + depreciation

print(f"EBIT x (1 - T)          : ${after_tax_ebit:,.0f}")
print(f"OCF = above + Dep.      : ${OCF:,.0f}")
````

---
## 2. Capital Spending — CapEx

**CapEx** is cash spent on long-term assets: machinery, buildings, vehicles, computers, production equipment, technology infrastructure.

```text
Company buys machinery
        |
     $20,000
        |
     Cash OUT
        |
      CapEx
```

If the company spends \$20,000 on machinery:

$$\boxed{CapEx = \$20{,}000}$$

This reduces the cash available to the company.

---
## 3. Change in Net Working Capital — ΔNWC

**Net Working Capital (NWC)** is the difference between operating current assets and operating current liabilities:

$$\boxed{NWC = \text{Current Assets} - \text{Current Liabilities}}$$

In corporate finance we usually focus on the *operating* pieces:

| Operating current assets | Operating current liabilities |
|---|---|
| Accounts receivable | Accounts payable |
| Inventory | Accrued operating expenses |

**Example**

| | NWC |
|---|---:|
| Beginning of year | \$20,000 |
| End of year | \$27,000 |

````python
beginning_nwc = 20_000
ending_nwc = 27_000
delta_nwc = ending_nwc - beginning_nwc

print(f"Delta NWC = {ending_nwc:,} - {beginning_nwc:,} = ${delta_nwc:,.0f}")
print("More cash tied up in working capital -> reduces free cash flow.")
````

---
## 4. Free Cash Flow — FCF

Combining the three pieces:

$$\boxed{FCF = OCF - CapEx - \Delta NWC}$$

- **OCF** — cash generated by operations
- **CapEx** — cash invested in long-term assets
- **ΔNWC** — additional cash tied up in working capital

### Complete Example

| Item | Amount |
|---|---:|
| EBIT | \$30,000 |
| Tax rate | 25% |
| Depreciation | \$5,000 |
| Capital Spending | \$10,000 |
| Beginning NWC | \$20,000 |
| Ending NWC | \$25,000 |

````python
EBIT = 30_000
tax_rate = 0.25
depreciation = 5_000
capex = 10_000
beginning_nwc = 20_000
ending_nwc = 25_000

# Step 1 - OCF
OCF = EBIT * (1 - tax_rate) + depreciation

# Step 2 - Delta NWC
delta_nwc = ending_nwc - beginning_nwc

# Step 3 - FCF
FCF = OCF - capex - delta_nwc

print(f"OCF  = EBIT(1-T) + Dep.        = ${OCF:,.0f}")
print(f"dNWC = Ending - Beginning NWC  = ${delta_nwc:,.0f}")
print(f"FCF  = OCF - CapEx - dNWC      = ${FCF:,.0f}")
````

$$\boxed{FCF = \$12{,}500}$$

### The Big Picture

![Cash Flow Diagram](q.png)

$$\boxed{\textbf{FCF = OCF} - \textbf{CapEx} - \boldsymbol{\Delta}\textbf{NWC}}$$

Or conceptually: **cash generated by the business − cash invested in long-term assets − cash tied up in operations = Free Cash Flow.**

### Intuition

| Driver | Change | Effect on FCF |
|---|:---:|:---:|
| OCF | ↑ | ↑ |
| CapEx | ↑ | ↓ |
| NWC | ↑ | ↓ |
| NWC | ↓ | ↑ |
