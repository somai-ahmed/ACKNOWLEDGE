## 1 Comments — Making Your Code Readable

Comments are notes you leave for yourself (and other humans) inside your code. Python completely ignores them during execution, but they are **essential** for writing maintainable, professional code.

### Why use comments?
- Explain *why* you wrote a particular piece of code
- Temporarily disable code during debugging
- Document complex logic for your future self

### Types of Comments in Python
| Type | Syntax | Use Case |
|------|--------|----------|
| Single-line | `# comment` | Quick notes |
| Inline | `code  # comment` | Explain a specific line |
| Multi-line | `"""text"""` | Docstrings & detailed docs |

````python
# This is a single-line comment
print("Hello, World!")  # This inline comment explains the line

"""
This is a multi-line string (docstring).
Often used at the beginning of functions or modules
to document what they do.
"""

# Try uncommenting the line below to see what happens!
# print("This line is currently disabled")
````

## 2 Variable Assignment — Storing Data

Variables are like labeled containers that hold data. In Python, you don't need to declare the type — Python figures it out automatically!

### Naming Rules
- Must start with a letter or underscore (`_`)
- Can contain letters, numbers, and underscores
- Case-sensitive (`name` ≠ `Name`)
- Avoid Python reserved words (`if`, `for`, `class`, etc.)

### Best Practices
- Use descriptive names: `user_age` instead of `x`
- Use `snake_case` for variables and functions
- Use `UPPER_CASE` for constants

````python
# Basic assignments
name = "John"          # String
age = 25               # Integer
height = 5.9           # Float
is_student = True      # Boolean

# Multiple assignments in one line
x, y, z = 10, 20, 30

# Swapping values (Python magic!)
a, b = 5, 10
a, b = b, a  # Now a=10, b=5

print(f"Name: {name}, Age: {age}, Height: {height}")
print(f"Is student? {is_student}")
print(f"Swapped: a={a}, b={b}")
````

## 3 Data Types — The DNA of Your Data

Every piece of data in Python has a type. Understanding these types is crucial because different types support different operations.

| Type | Description | Example |
|------|-------------|---------|
| `int` | Whole numbers (positive or negative) | `42`, `-7` |
| `float` | Decimal numbers | `3.14`, `-0.5` |
| `bool` | Logical values | `True`, `False` |
| `str` | Text (sequences of characters) | `"Hello"` |

> 🔍 You can check any value's type using the `type()` function.

````python
# Exploring different data types
integer_value = 7
float_value = 12.4
boolean_true = True
boolean_false = False
string_value = "Hello, Python!"

# Let's inspect their types
print(f"{integer_value} is of type: {type(integer_value).__name__}")
print(f"{float_value} is of type: {type(float_value).__name__}")
print(f"{boolean_true} is of type: {type(boolean_true).__name__}")
print(f"{string_value} is of type: {type(string_value).__name__}")

# Type conversion (casting)
num_str = "100"
num_int = int(num_str)      # String → Integer
num_float = float(num_str)  # String → Float

print(f"Converted '{num_str}' to int: {num_int + 50}")
print(f"Converted '{num_str}' to float: {num_float + 0.5}")
````

## 4️⃣ The `print()` Function — Your Window to the World

`print()` is one of the most frequently used functions in Python. It outputs text or variables to the console, which is invaluable for debugging and displaying results.

### Advanced Features
- **f-strings**: Embed variables directly using `f"...{var}..."`
- **sep parameter**: Customize what separates multiple values
- **end parameter**: Control what appears at the end (default is newline `\n`)

````python
# Basic printing
print("Hello, world!")

# Printing variables
a = 10
b = 20
print(a + b)  # Prints: 30

# f-strings (formatted string literals) — Modern & clean!
name = "Alice"
age = 30
print(f"My name is {name} and I am {age} years old.")

# Using sep and end parameters
print("Apple", "Banana", "Cherry", sep=" → ", end=" | DONE\n")

# Pretty printing multiple values
print(f"{'Item':<15} {'Price':>10}")
print(f"{'Coffee':<15} {'$3.50':>10}")
print(f"{'Sandwich':<15} {'$7.25':>10}")
````

## 4 Python Operators — The Math Magicians

Operators perform operations on values and variables. Python supports a rich set of operators for arithmetic, comparison, logic, and more.

### Arithmetic Operators
| Operator | Name | Description | Example | Result |
|----------|------|-------------|---------|--------|
| `+` | Addition | Adds two values | `9 + 4` | `13` |
| `-` | Subtraction | Subtracts one from another | `9 - 4` | `5` |
| `*` | Multiplication | Multiplies two values | `9 * 4` | `36` |
| `/` | Division | Divides, returns float | `9 / 4` | `2.25` |
| `//` | Floor Division | Divides, returns integer quotient | `9 // 4` | `2` |
| `%` | Modulo | Returns remainder | `9 % 4` | `1` |
| `**` | Exponentiation | Raises to power | `2 ** 3` | `8` |

> **Pro Tip**: `//` is great when you need whole numbers (e.g., "how many full boxes fit?"). `%` tells you what's left over (e.g., "how many items remain?").

````python
x = 9
y = 4

print("📊 Arithmetic Operations Demo\n" + "="*40)

result_add = x + y   # Addition
result_sub = x - y   # Subtraction
result_mul = x * y   # Multiplication
result_div = x / y   # Division (always float!)
result_fdiv = x // y # Floor Division
result_mod = x % y   # Modulo (remainder)
result_exp = x ** y  # Exponentiation

print(f"{x} + {y} = {result_add}")
print(f"{x} - {y} = {result_sub}")
print(f"{x} * {y} = {result_mul}")
print(f"{x} / {y} = {result_div}  ← Notice the decimal!")
print(f"{x} // {y} = {result_fdiv}  ← Whole number only")
print(f"{x} % {y} = {result_mod}  ← What's left after division")
print(f"{x} ** {y} = {result_exp}  ← {x} to the power of {y}")

# Real-world example: Time conversion
total_seconds = 3665
hours = total_seconds // 3600
remaining = total_seconds % 3600
minutes = remaining // 60
seconds = remaining % 60
print(f"\n {total_seconds}s = {hours}h {minutes}m {seconds}s")
````

## 5 String Methods — Transforming Text

Strings in Python are **immutable** (they can't be changed after creation), but Python provides powerful methods that return *new* transformed strings.

### Essential String Methods
| Method | What It Does | Returns |
|--------|--------------|---------|
| `len(s)` | Counts total characters | `int` |
| `s.upper()` | Converts to ALL CAPS | `str` |
| `s.lower()` | Converts to all lowercase | `str` |
| `s.strip()` | Removes whitespace from edges | `str` |

> ⚠️ Remember: These methods return a NEW string. The original string stays unchanged!

````python
my_string = "  Hello, Python!  "

print(f"Original: '{my_string}'")
print(f"Length: {len(my_string)} characters")
print(f"Uppercase: '{my_string.upper()}'")
print(f"Lowercase: '{my_string.lower()}'")
print(f"Stripped: '{my_string.strip()}'")

# Verify immutability
print(f"\nAfter all operations, original is still: '{my_string}'")

# Practical use case: Normalizing user input
user_input = "  YeS  "
clean_input = user_input.strip().lower()
print(f"\nUser said: '{user_input}' → Normalized: '{clean_input}'")
print(f"Is it 'yes'? {clean_input == 'yes'}")
````

## 7 Indexing & Slicing — Precision Text Extraction

Python strings are sequences of characters, and each character has a position (index). You can access individual characters or extract substrings using slicing.

### Indexing Visualization
```
String:  H   e   l   l   o
Index:   0   1   2   3   4
Reverse: -5  -4  -3  -2  -1
```

### Slicing Syntax
```python
substring = string[start:end:step]
```
- `start`: Index to begin (inclusive)
- `end`: Index to stop (exclusive)
- `step`: How many characters to skip (optional, default is 1)

````python
my_string = "Hello, World!"

print(f"String: '{my_string}'")
print(f"Length: {len(my_string)}")
print("\n Indexing:")
print(f"First character: '{my_string[0]}'")
print(f"Last character: '{my_string[-1]}'")
print(f"7th character: '{my_string[6]}'")

print("\n Slicing:")
print(f"First 5 chars: '{my_string[0:5]}'")
print(f"From index 7 to end: '{my_string[7:]}'")
print(f"Up to index 5: '{my_string[:5]}'")
print(f"Every 2nd char: '{my_string[::2]}'")
print(f"Reversed: '{my_string[::-1]}' ← Python magic!")

# Extracting a domain from email
email = "student@university.edu"
at_index = email.index("@")
domain = email[at_index + 1:]
print(f"\n📧 Email: {email}")
print(f"Domain: {domain}")
````

## 8 `replace()` & `split()` — Text Manipulation Powerhouses

These two methods are incredibly useful for processing and transforming text data.

### `replace(old, new, count)`
Replaces occurrences of a substring with another substring.
- `old`: The substring to find
- `new`: The substring to replace it with
- `count`: (Optional) Maximum number of replacements

### `split(sep, maxsplit)`
Breaks a string into a list of substrings based on a delimiter.
- `sep`: The delimiter (default is whitespace)
- `maxsplit`: (Optional) Maximum number of splits

````python
# replace() examples
text = "Hello, World! Hello, Python!"
print(f"Original: '{text}'")

new_text = text.replace("Hello", "Hi")
print(f"Replace all: '{new_text}'")

limited_replace = text.replace("Hello", "Hi", 1)
print(f"Replace once: '{limited_replace}'")

# split() examples
csv_line = "Alice,25,Engineer,New York"
print(f"\nCSV line: '{csv_line}'")

parts = csv_line.split(",")
print(f"Split by comma: {parts}")

sentence = "The quick brown fox jumps"
words = sentence.split()  # Default: split by whitespace
print(f"\nSentence: '{sentence}'")
print(f"Words: {words}")
print(f"Word count: {len(words)}")

# Practical: Parsing a file path
path = "/home/user/documents/report.pdf"
components = path.split("/")
filename = components[-1]
extension = filename.split(".")[-1]
print(f"\n Path: {path}")
print(f"Filename: {filename}")
print(f"Extension: {extension}")
````

## 9 String Concatenation — Building Text Together

Concatenation is the process of joining strings together. Python offers multiple ways to combine text.

### Methods
| Method | Syntax | Best For |
|--------|--------|----------|
| `+` operator | `s1 + s2` | Few strings |
| `join()` | `sep.join(list)` | Many strings |
| f-strings | `f"{var}"` | Mixed text & variables |

> ⚠️ Using `+` in a loop with many strings is inefficient. Use `join()` or f-strings instead!

````python
# Using + operator
greeting = "Hello"
name = "John"
result = greeting + ", " + name + "!"
print(f"Using +: '{result}'")

# Using join() — efficient for lists
words = ["Python", "is", "awesome"]
sentence = " ".join(words)
print(f"Using join(): '{sentence}'")

# Using f-strings — modern and readable
language = "Python"
version = 3.12
message = f"I am learning {language} version {version}!"
print(f"Using f-string: '{message}'")

# Building a formatted report
items = ["Apple", "Banana", "Cherry"]
report = "Shopping List:\n" + "\n".join(f"  • {item}" for item in items)
print(f"\n{report}")
````
