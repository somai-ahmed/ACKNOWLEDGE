# Quick Start to R Language

This notebook introduces the basic installation and first commands in **R**.

## 1. Install the R environment

Download and install **R for Windows** from CRAN:

- [CRAN — R for Windows](https://cran.r-project.org/bin/windows/base/)
- [Download R-4.6.1 for Windows](https://cran.r-project.org/bin/windows/base/R-4.6.1-win.exe)

After downloading the installer:

1. Run the installer.
2. Follow the installation steps.
3. Wait for the installation to finish.
4. Start **R**.

## 2. Test the R environment

Once R is open, you can enter commands in the console.

### Basic arithmetic

```r
5 + 7
```

### Absolute value

The `abs()` function returns the absolute value of a number.

```r
abs(-17)
```

## 3. Variables

In R, `<-` is commonly used to assign a value to a variable.

> **Important:** `x <- 12` assigns `12` to `x`.  
> Writing `x <- -12` assigns `-12` to `x`.

```r
x <- 12
x
```

```r
x <- -12
x
```

## 4. Vectors

The `c()` function combines several values into a vector.

```r
y <- c(-12, 6, 0, -1)
y
```

### Absolute values of a vector

R can apply `abs()` to every element of the vector.

```r
abs(y)
```

## Quick recap

| Concept | R syntax | Example |
|---|---|---|
| Addition | `+` | `5 + 7` |
| Absolute value | `abs()` | `abs(-17)` |
| Assignment | `<-` | `x <- 12` |
| Vector creation | `c()` | `c(-12, 6, 0, -1)` |
| Display a variable | Variable name | `x` |
| Apply a function to a vector | Function + vector | `abs(y)` |
