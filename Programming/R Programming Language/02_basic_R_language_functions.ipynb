## 7. Missing values: `NA`

In R, `NA` means that a value is missing or unknown.

If a numeric vector contains an `NA`, a calculation such as `mean()` normally returns `NA` because R cannot calculate the complete result from an incomplete vector.

The argument `na.rm = TRUE` means **remove/ignore missing `NA` values when performing the calculation**.

Example:

```r
mean(c(8, 7.5, NA, 6.5))
```

returns `NA`, while:

```r
mean(c(8, 7.5, NA, 6.5), na.rm = TRUE)
```

calculates the mean of the available numbers.

````r
mean(c(8, 7.5, NA, 6.5))

mean(c(8, 7.5, NA, 6.5), na.rm = TRUE)
````

## 8. The important problem we encountered with `imdb`

In the original work, the following command produced a warning:

```r
mean(Scooby_Doo_Completed$imdb, na.rm = TRUE)
```

The warning was:

```text
argument is not numeric or logical: returning NA
```

At first this can look like an `NA` problem. However, the real issue was the **data type**.

Checking the column showed:

```r
class(Scooby_Doo_Completed$imdb)
```

with result:

```text
[1] "character"
```

So R was treating the IMDb values as text rather than numbers.

````r
class(Scooby_Doo_Completed$imdb)

typeof(Scooby_Doo_Completed$imdb)
````

## 9. Character vs numeric

These are different in R:

```text
"7.5"   -> character (text)
7.5     -> numeric (number)
```

A function such as `mean()` needs a numeric or logical vector. Therefore, if `imdb` is character, we need to convert it.

````r
# Convert the IMDb column from character to numeric
Scooby_Doo_Completed$imdb <- as.numeric(Scooby_Doo_Completed$imdb)

# Check the new type
class(Scooby_Doo_Completed$imdb)
````

## 10. A useful distinction: two different problems

### Problem A — numeric column containing `NA`

```r
mean(x, na.rm = TRUE)
```

This is the situation where `na.rm = TRUE` solves the problem.

### Problem B — column is `character`

```r
class(x)
# "character"
```

Here, `na.rm = TRUE` does **not** convert text into numbers. The column must first be converted:

```r
x <- as.numeric(x)
```

Then you can use:

```r
mean(x, na.rm = TRUE)
```
