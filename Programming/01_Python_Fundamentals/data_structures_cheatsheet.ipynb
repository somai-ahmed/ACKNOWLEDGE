## List
A list is a built-in data type that represents an ordered and mutable collection of elements. Lists are enclosed in square brackets `[]` and elements are separated by commas.

### Creating a list

```python
fruits = ["apple", "banana", "orange", "mango"]
print(fruits)
```

### Indexing
Indexing in a list allows you to access individual elements by their position. In Python, indexing starts from `0` for the first element and goes up to `length_of_list - 1`.

```python
my_list = [10, 20, 30, 40, 50]
print(my_list[0])   # Output: 10 (accessing the first element)
print(my_list[-1])  # Output: 50 (accessing the last element using negative indexing)
```

### Modifying a list
You can use indexing to modify or assign new values to specific elements in the list.

```python
my_list = [10, 20, 30, 40, 50]
my_list[1] = 25  # Modifying the second element
print(my_list)   # Output: [10, 25, 30, 40, 50]
```

### Slicing
You can use slicing to access a range of elements from a list.

**Syntax:** `list_name[start:end:step]`

```python
my_list = [1, 2, 3, 4, 5]
print(my_list[1:4])   # Output: [2, 3, 4] (elements from index 1 to 3)
print(my_list[:3])    # Output: [1, 2, 3] (elements from the beginning up to index 2)
print(my_list[2:])    # Output: [3, 4, 5] (elements from index 2 to the end)
print(my_list[::2])   # Output: [1, 3, 5] (every second element)
```

### `append()`
The `append()` method is used to add an element to the end of a list.

**Syntax:** `list_name.append(element)`

```python
fruits = ["apple", "banana", "orange"]
fruits.append("mango")
print(fruits)
```

### `extend()`
The `extend()` method is used to add multiple elements to a list. It takes an iterable (such as another list, tuple, or string) and appends each element of the iterable to the original list.

**Syntax:** `list_name.extend(iterable)`

```python
fruits = ["apple", "banana", "orange"]
more_fruits = ["mango", "grape"]
fruits.extend(more_fruits)
print(fruits)
```

### `insert()`
The `insert()` method is used to insert an element at a specific index.

**Syntax:** `list_name.insert(index, element)`

```python
my_list = [1, 2, 3, 4, 5]
my_list.insert(2, 6)
print(my_list)
```

### `remove()`
The `remove()` method removes the first occurrence of the specified value.

```python
my_list = [10, 20, 30, 40, 50]
my_list.remove(30)  # Removes the element 30
print(my_list)      # Output: [10, 20, 40, 50]
```

### `pop()`
The `pop()` method removes and returns the element at the specified index. If no index is provided, it removes and returns the last element by default.

```python
# Example 1: Remove by index
my_list = [10, 20, 30, 40, 50]
removed_element = my_list.pop(2)  # Removes and returns the element at index 2
print(removed_element)            # Output: 30
print(my_list)                    # Output: [10, 20, 40, 50]

# Example 2: Remove last element
my_list = [10, 20, 30, 40, 50]
removed_element = my_list.pop()   # Removes and returns the last element
print(removed_element)            # Output: 50
print(my_list)                    # Output: [10, 20, 30, 40]
```

### `del`
The `del` statement is used to remove an element from a list by its index.

```python
my_list = [10, 20, 30, 40, 50]
del my_list[2]   # Removes the element at index 2
print(my_list)   # Output: [10, 20, 40, 50]
```

### `copy()`
The `copy()` method is used to create a shallow copy of a list.

```python
my_list = [1, 2, 3, 4, 5]
new_list = my_list.copy()
print(new_list)  # Output: [1, 2, 3, 4, 5]
```

### `count()`
The `count()` method is used to count the number of occurrences of a specific element in a list.

```python
my_list = [1, 2, 2, 3, 4, 2, 5, 2]
count = my_list.count(2)
print(count)  # Output: 4
```

### `reverse()`
The `reverse()` method is used to reverse the order of elements in a list.

```python
my_list = [1, 2, 3, 4, 5]
my_list.reverse()
print(my_list)  # Output: [5, 4, 3, 2, 1]
```

### `sort()`
The `sort()` method is used to sort the elements of a list in ascending order. Pass `reverse=True` to sort in descending order.

```python
# Example 1: Ascending order
my_list = [5, 2, 8, 1, 9]
my_list.sort()
print(my_list)  # Output: [1, 2, 5, 8, 9]

# Example 2: Descending order
my_list = [5, 2, 8, 1, 9]
my_list.sort(reverse=True)
print(my_list)  # Output: [9, 8, 5, 2, 1]
```

## Tuple
A tuple is an ordered, immutable collection of elements enclosed in parentheses `()`.

### `count()`
The `count()` method counts how many times a specified element appears in the tuple.

**Syntax:** `tuple.count(value)`

```python
fruits = ("apple", "banana", "apple", "orange")
print(fruits.count("apple"))  # Output: 2
```

### `index()`
The `index()` method finds the first occurrence of a specified value and returns its position (index). If the value is not found, it raises a `ValueError`.

**Syntax:** `tuple.index(value)`

```python
fruits = ("apple", "banana", "orange", "apple")
print(fruits.index("apple"))  # Output: 0
```

### `sum()`
The `sum()` function calculates the sum of all elements in a tuple, provided the elements are numeric (integers or floats).

**Syntax:** `sum(tuple)`

```python
numbers = (10, 20, 5, 30)
print(sum(numbers))  # Output: 65
```

### `min()` and `max()`
Find the smallest (`min()`) or largest (`max()`) element in a tuple.

```python
numbers = (10, 20, 5, 30)
print(min(numbers))  # Output: 5
print(max(numbers))  # Output: 30
```

### `len()`
Get the number of elements in the tuple using `len()`.

**Syntax:** `len(tuple)`

```python
fruits = ("apple", "banana", "orange")
print(len(fruits))  # Output: 3
```
