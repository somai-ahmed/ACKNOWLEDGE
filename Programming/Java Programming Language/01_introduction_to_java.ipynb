## I. Object

An **object** is an entity from the real world. It has 2 properties:

| Property | Meaning |
|---|---|
| **State** | Attributes |
| **Behavior** | Methods |

**Example: ARTICLE**

| Attributes | Methods |
|---|---|
| CODE | Enter stock |
| LABEL | Exit stock |
| QUANTITE | Calculate |
| HT PRICE (Price including tax) | |

---
## Note: Structured Programming vs. Object-Oriented Programming

| Structured Programming | Object-Oriented Programming (OOP) |
|---|---|
| What should the program do? | What does the program operate on? |
| Reasoning through **processing** | Reasoning through **objects** |

## II. Classes

A class is an **object type**. It defines:
- the attributes of its objects
- the methods

```python
CLASS ARTICLE {
    //ATRIBUTS 
    int code ;
    string label ;
    int qte ;
    double PRICE_HT ;

    //METHODS

    void enter_stock(int n){
        qte = qte +n ;          //qte+=n//
    }

    void exit_stock(int n){
        qte = qte -n ;          //qte-=n//
    }
    
    double calculTTC(){
        return (PRICE_HT+PRICE_HT*0.1) ;
    }
}

main{
    ARTICLE a1

    a1.enter_stock(50);
    double res = calculTTC();
}
```
